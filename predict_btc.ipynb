```python
import yfinance as yf
```

```python
btc = yf.Ticker("BTC-USD")
```

```python
btc = btc.history(period="max")
```

```python
btc
```

```python
btc.index
```

```python
btc.plot.line(y="Close", use_index=True)
```

```python
btc["Tomorrow"] = btc["Close"].shift(-1)
```

```python
btc
```

```python
btc["Target"] = (btc["Tomorrow"] > btc["Close"]).astype(int)
```

```python
btc
```

```python
btc = btc.loc["2016-01-01":].copy()
```

```python
btc
```

```python
train = btc.iloc[:-100]
test = btc.iloc[-100:]

predictors = ["Close", "Volume", "Open", "High", "Low", "Returns"]
```

```python
scale_pos_weight = (train["Target"] == 0).sum() / (train["Target"] == 1).sum() #calculates a weight to handle class imbalance
```

```python
from xgboost import XGBClassifier
model1 = XGBClassifier(
    n_estimators=300,
    max_depth=15,
    learning_rate=0.1,
    subsample=0.9,
    colsample_bytree=0.8,
    scale_pos_weight=scale_pos_weight,
    use_label_encoder=False,
    eval_metric='logloss',
    random_state=1
)

model1.fit(train[predictors], train["Target"])
```

```python
from sklearn.metrics import precision_score

predictions = model1.predict(test[predictors])
```

```python
import pandas as pd 

predictions = pd.Series(predictions, index=test.index)
```

```python
precision_score(test["Target"], predictions)
```

```python
combined = pd.concat([test["Target"], predictions], axis=1)
```

```python
combined.plot()
```

```python
def predict(train, test, predictors, model): 
        model.fit(train[predictors], train["Target"])
        predictions = model.predict(test[predictors])
        predictions = pd.Series(predictions, index=test.index, name = "Predictions")
        combined = pd.concat([test["Target"], predictions], axis=1)
        return combined
```

```python
def backtest(data, model, predictors, start=2500, step=250):
    all_predictions = []
    
    for i in range(start, data.shape[0], step):
        train = data.iloc[0:i].copy()
        test = data.iloc[i:(i+step)].copy()
        predictions = predict(train, test, predictors, model)
        all_predictions.append(predictions)
    return pd.concat(all_predictions)
```

```python
predictions = backtest(btc, model1, predictors)
```

```python
predictions["Predictions"].value_counts()
```

```python
precision_score(predictions["Target"], predictions["Predictions"])
```

```python
predictions["Target"].value_counts() / predictions.shape[0]
```

```python
#improve basic model
```

```python
horizons = [2, 5, 60, 250, 1000]

new_predictors = []

for horizon in horizons:
    rolling_averages = btc.rolling(horizon).mean()
    
    ratio_column = f"Close Ratio_{horizon}"
    btc[ratio_column] = btc["Close"] / rolling_averages["Close"]
    
    trend_column = f"Trend_{horizon}"
    btc[trend_column] = btc.shift(1).rolling(horizon).sum()["Target"]
    
    new_predictors += [ratio_column, trend_column]
```

```python
btc = btc.dropna()
```

```python
btc
```

```python
model = XGBClassifier(
    n_estimators=500,
    max_depth=15,
    learning_rate=0.1,
    subsample=0.9,
    colsample_bytree=0.8,
    scale_pos_weight=scale_pos_weight,
    use_label_encoder=False,
    eval_metric='logloss',
    random_state=1
)
```

```python
def predict(train, test, predictors, model): 
        model.fit(train[predictors], train["Target"])
        predictions = model.predict_proba(test[predictors])[:,1]
        predictions[predictions >= .6] = 1
        predictions[predictions < .6] = 0
        predictions = pd.Series(predictions, index=test.index, name = "Predictions")
        combined = pd.concat([test["Target"], predictions], axis=1)
        return combined
```

```python
predictions = backtest(btc, model, new_predictors, start=500, step=250)
```

```python
predictions["Predictions"].value_counts()
```

```python
precision_score(predictions["Target"], predictions["Predictions"])
```
